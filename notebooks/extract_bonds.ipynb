# Data Extraction: Bonds (10 ETFs)

| ETF | Full Name | Focus |
|-----|-----------|-------|
| SHY | iShares 1-3 Year Treasury Bond ETF | Short term US Treasuries |
| IEF | iShares 7-10 Year Treasury Bond ETF | Medium term US Treasuries |
| TLT | iShares 20+ Year Treasury Bond ETF | Long term US Treasuries |
| LQD | iShares Investment Grade Corporate Bond ETF | High quality corporate bonds |
| HYG | iShares High Yield Corporate Bond ETF | Junk / high yield corporate bonds |
| MUB | iShares National Municipal Bond ETF | Tax exempt municipal bonds |
| TIP | iShares TIPS Bond ETF | Inflation protected Treasuries |
| EMB | iShares Emerging Market Bond ETF | Emerging market government bonds |
| AGG | iShares Core US Aggregate Bond ETF | Broad US bond market |
| BND | Vanguard Total Bond Market ETF | Broad US bond market |

```python
import yfinance as yf
import pandas as pd
import os
import time
```

```python
START = "2011-01-01"
END   = "2025-12-31"

TICKERS = {
    "SHY": "SHY",   # iShares 1-3 Year Treasury Bond ETF
    "IEF": "IEF",   # iShares 7-10 Year Treasury Bond ETF
    "TLT": "TLT",   # iShares 20+ Year Treasury Bond ETF
    "LQD": "LQD",   # iShares Investment Grade Corporate Bond ETF
    "HYG": "HYG",   # iShares High Yield Corporate Bond ETF
    "MUB": "MUB",   # iShares National Municipal Bond ETF
    "TIP": "TIP",   # iShares TIPS Bond ETF
    "EMB": "EMB",   # iShares Emerging Market Bond ETF
    "AGG": "AGG",   # iShares Core US Aggregate Bond ETF
    "BND": "BND",   # Vanguard Total Bond Market ETF
}

```

```python
raw = {}
for name, ticker in TICKERS.items():
    print(f"Downloading {name} ({ticker})...")
    for attempt in range(1, 6):
        try:
            df = yf.download(
                ticker,
                start=START,
                end=END,
                auto_adjust=True,
                progress=False,
                threads=False,
            )
            if not df.empty:
                break
        except Exception as e:
            df = pd.DataFrame()
            print(f"  Error: {e}")
        wait = attempt * 30
        print(f"  Retrying in {wait}s (attempt {attempt}/5)...")
        time.sleep(wait)

    if df.empty:
        print(f"  FAILED: {name}")
        continue

    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    df.index = pd.to_datetime(df.index)
    raw[name] = df
    print(f"  {len(df)} rows  |  {df.index[0].date()} -> {df.index[-1].date()}")
    time.sleep(2)
```

```python
for name, df in raw.items():
    print(f"\n--- {name} ---")
    print(df.shape)
    print(df.isna().sum())
    display(df.tail())
```

```python
for name, df in raw.items():
    print(f"\n--- {name} ---")
    display(df["Close"].describe())
```

```python
os.makedirs("../data/bonds", exist_ok=True)

for name, df in raw.items():
    out = f"../data/bonds/raw_{name.lower()}.csv"
    df.to_csv(out)
    print(f"Saved {out}")
```
